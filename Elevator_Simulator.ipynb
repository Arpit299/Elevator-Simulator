### Elevator Simulator

```python
import time

class Elevator:
    def __init__(self, name="Elevator 1", floors=10):
        self.name = name
        self.current_floor = 1
        self.total_floors = floors
        self.is_door_open = False

    def open_doors(self):
        if not self.is_door_open:
            print(f"[{self.name}] Opening doors at floor {self.current_floor}...")
            self.is_door_open = True

    def close_doors(self):
        if self.is_door_open:
            print(f"[{self.name}] Closing doors...")
            self.is_door_open = False

    def move_to_floor(self, target_floor):
        if target_floor < 1 or target_floor > self.total_floors:
            print(f"Invalid floor: {target_floor}")
            return

        if target_floor == self.current_floor:
            print(f"[{self.name}] Already at floor {target_floor}.")
            self.open_doors()
            return

        self.close_doors()
        direction = 1 if target_floor > self.current_floor else -1

        print(f"[{self.name}] Moving from floor {self.current_floor} to {target_floor}...")
        while self.current_floor != target_floor:
            self.current_floor += direction
            print(f"... passing floor {self.current_floor}")
            time.sleep(0.5)

        print(f"[{self.name}] Arrived at floor {self.current_floor}.")
        self.open_doors()

my_elevator = Elevator(floors=5)
my_elevator.move_to_floor(4)
my_elevator.move_to_floor(1)
```
